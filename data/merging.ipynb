```python
import pandas as pd
```

```python
crop_df = pd.read_csv("crop_yield.csv")
rainfall_df = pd.read_csv("rainfall in india 1901-2015.csv")
```

```python
print(crop_df.columns)
print(rainfall_df.columns)
```

```python
rainfall_df.rename(columns={
    "YEAR": "Crop_Year",
    "SUBDIVISION": "State",
    "ANNUAL": "Annual_Rainfall"
}, inplace=True)
```

```python
rainfall_df = rainfall_df[["State", "Crop_Year", "Annual_Rainfall"]]
```

```python
crop_df["State"] = crop_df["State"].str.strip().str.upper()
rainfall_df["State"] = rainfall_df["State"].str.strip().str.upper()
```

```python
print(set(crop_df["State"]).intersection(set(rainfall_df["State"])))
```

```python
merged_df = pd.merge(
    crop_df,
    rainfall_df,
    on=["State", "Crop_Year"],
    how="inner"
)
```

```python
print(merged_df.head())
print(merged_df.shape)
```

```python
merged_df.to_csv("merged_dataset.csv", index=False)
```
